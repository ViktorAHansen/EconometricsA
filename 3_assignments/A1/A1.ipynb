```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
#PROBLEM 1

data = pd.read_stata("A1_kommune.dta")
data['const'] = 1
data['ltaxrev'] = np.log(data['taxrev'])

# Deskriptiv statistik
descriptive_stats = data.describe()

# Display
display(descriptive_stats)
```

```python
#PROBLEM 2
def OLS(X, y):
    X = X.values
    y = y.values
    n = X.shape[0]
    k = X.shape[1] - 1

    deltahat = np.linalg.inv(X.T @ X) @ X.T @ y
    yhat = X @ deltahat
    uhat = y - yhat

    ybar = y.mean()

    SST = np.sum((y - ybar)**2)
    SSE = np.sum((yhat - ybar)**2)
    SSR = np.sum(uhat**2)

    R2 = SSE / SST
    
    sigma2 = np.dot(uhat, uhat) * 1/(n-k-1)

    var_deltahat = sigma2 * np.linalg.inv(X.T @ X)
    se_deltahat = np.sqrt(np.diag(var_deltahat))

    return {
        'deltahat': deltahat,
        'se_deltahat': se_deltahat,
        'yhat': yhat,
        'uhat': uhat,
        'sigma2': sigma2,
        'R2': R2,
        'SST': SST,
        'SSE': SSE,
        'SSR': SSR,
    }


OLS(data[['const', 'taxrate']], data['ltaxrev'])
```

```python
import statsmodels.api as sm

model = sm.OLS(data['ltaxrev'], data[['const', 'taxrate']]).fit()

print(model.summary())
```
